```python

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.001


def get_next_state(state, action):
    transitions = {
        'S0': {
            'Up': ('S0', -1),
            'Down': ('S1', -1),
            'Left': ('S0', -1),
            'Right': ('S0', -1)
        },
        'S1': {
            'Up': ('S0', -1),
            'Down': ('S1', -1),
            'Left': ('S1', -1),
            'Right': ('G', 100)
        }
    }
    return transitions[state][action]



policy = {
    'S0': 'Up',
    'S1': 'Up'
}


V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}


def policy_evaluation():
    while True:
        delta = 0

        for state in ['S0', 'S1']:
            next_state, reward = get_next_state(state, policy[state])

            new_value = reward + gamma * V[next_state]

            delta = max(delta, abs(V[state] - new_value))
            V[state] = new_value

        if delta < theta:
            break


while True:
    policy_evaluation()

    policy_stable = True

    for state in ['S0', 'S1']:
        old_action = policy[state]

        action_values = {}

        for action in actions:
            next_state, reward = get_next_state(state, action)
            action_values[action] = reward + gamma * V[next_state]

        best_action = max(action_values, key=action_values.get)

        if old_action != best_action:
            policy_stable = False

        policy[state] = best_action

    if policy_stable:
        break


# Final Results
print("Optimal State Values:")
for state in states:
    print(state, ":", round(V[state], 2))

print("\nOptimal Policy:")
for state in ['S0', 'S1']:
    print(state, ":", policy[state])
```
