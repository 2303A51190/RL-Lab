```python
import math

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of machines and number of plays
M = 3
N_play = 10

# Part B - Initialize N, R, UCB and Q
N = [0, 0, 0]          # Number of times each machine is selected
R = [0, 0, 0]          # Total reward of each machine
Q = [0.0, 0.0, 0.0]    # Estimated Q-value
UCB = [0.0, 0.0, 0.0]   # UCB values

total_reward = 0

print("UCB Multi-Armed Bandit")
print("----------------------")

# ------------------------------------------------
# INITIAL TEST - Play each machine once
# ------------------------------------------------

for i in range(M):

    print("\nInitial Test")
    print("Play with Machine", machines[i])

    reward = float(input("Enter reward: "))

    # Update number of plays
    N[i] = N[i] + 1

    # Update total reward
    R[i] = R[i] + reward

    # Incremental sample-average Q-value
    Q[i] = Q[i] + (reward - Q[i]) / N[i]

    # Update total reward
    total_reward = total_reward + reward


# ------------------------------------------------
# UCB PLAYS - From Play 4 to Play 10
# ------------------------------------------------

for play in range(4, N_play + 1):

    # Part C - Calculate UCB
    for i in range(M):
        UCB[i] = Q[i] + 2 * math.sqrt(
            math.log(play) / N[i]
        )

    # Select machine with highest UCB
    selected = UCB.index(max(UCB))

    print("\nPlay", play)

    print("UCB Values:")
    for i in range(M):
        print(machines[i], "=", round(UCB[i], 4))

    print("Selected Machine:", machines[selected])

    # Part D - Get reward
    reward = float(input(
        "Enter reward for Machine " + machines[selected] + ": "
    ))

    # Part E - Update N
    N[selected] = N[selected] + 1

    # Update R
    R[selected] = R[selected] + reward

    # Update Q using incremental sample-average
    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    # Update total reward
    total_reward = total_reward + reward


# ------------------------------------------------
# Part F - Final Results
# ------------------------------------------------

# Calculate final UCB values
for i in range(M):
    UCB[i] = Q[i] + 2 * math.sqrt(
        math.log(N_play) / N[i]
    )

print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tR\tUCB\t\tQ")

for i in range(M):
    print(
        machines[i], "\t",
        N[i], "\t",
        R[i], "\t",
        round(UCB[i], 4), "\t\t",
        round(Q[i], 4)
    )

print("\nTotal Reward =", total_reward)
```
