```python
import random

# ---------------------------------------
# Multi-Armed Bandit
# Modified Epsilon-Greedy Algorithm
# ---------------------------------------

machines = ['A', 'B', 'C']
epsilon = 0.2
total_plays = 10

# ---------------------------------------
# Enter the rewards given in the question
# ---------------------------------------

rewards = {}

for machine in machines:
    print(f"\nEnter the 4 rewards for Machine {machine}:")
    rewards[machine] = []

    for i in range(4):
        value = float(input(f"Reward {i+1} for Machine {machine}: "))
        rewards[machine].append(value)

# Keep track of which reward is used next
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

# ---------------------------------------
# Initialize N, R and Q
# ---------------------------------------

N = {
    'A': 0,
    'B': 0,
    'C': 0
}

R = {
    'A': 0,
    'B': 0,
    'C': 0
}

Q = {
    'A': 0.0,
    'B': 0.0,
    'C': 0.0
}

total_reward = 0

# Fixed seed so the result can be reproduced
random.seed(42)

print("\n" + "=" * 55)
print("STARTING MULTI-ARMED BANDIT")
print("=" * 55)

# ---------------------------------------
# 10 Plays
# ---------------------------------------

for play in range(1, total_plays + 1):

    # Machines that still have rewards available
    available = [
        m for m in machines
        if reward_index[m] < len(rewards[m])
    ]

    # Stop if no rewards are available
    if len(available) == 0:
        print("\nNo more rewards available.")
        break

    # Find highest Q among available machines
    max_q = max(Q[m] for m in available)

    best_machines = [
        m for m in available
        if Q[m] == max_q
    ]

    highest_machine = best_machines[0]

    # Generate random number
    random_number = random.random()

    # -----------------------------------
    # Modified Epsilon-Greedy
    # -----------------------------------

    if random_number < epsilon:

        # Exploration:
        # choose randomly excluding highest-Q machine
        exploration_choices = [
            m for m in available
            if m != highest_machine
        ]

        # If only one machine is available
        if len(exploration_choices) == 0:
            selected = highest_machine
        else:
            selected = random.choice(exploration_choices)

        method = "Exploration"

    else:

        # Exploitation:
        # choose highest-Q machine
        selected = highest_machine
        method = "Exploitation"

    # -----------------------------------
    # Get the next given reward
    # -----------------------------------

    index = reward_index[selected]
    reward = rewards[selected][index]

    # Move to next reward for that machine
    reward_index[selected] += 1

    # -----------------------------------
    # Update N, R and Q
    # -----------------------------------

    N[selected] += 1
    R[selected] += reward

    # Incremental sample-average update
    Q[selected] = Q[selected] + (
        (reward - Q[selected]) / N[selected]
    )

    total_reward += reward

    # -----------------------------------
    # Display result of current play
    # -----------------------------------

    print(f"\nPlay {play}")
    print(f"Random number = {random_number:.4f}")
    print(f"Method = {method}")
    print(f"Selected Machine = {selected}")
    print(f"Reward received = {reward}")

    print(f"N = {N}")
    print(f"R = {R}")

    print("Q values:")
    print(f"A = {Q['A']:.2f}")
    print(f"B = {Q['B']:.2f}")
    print(f"C = {Q['C']:.2f}")

# ---------------------------------------
# Final Results
# ---------------------------------------

print("\n" + "=" * 55)
print("FINAL RESULTS")
print("=" * 55)

print("\nMachine\tN\tR\tQ")

for machine in machines:
    print(
        f"{machine}\t"
        f"{N[machine]}\t"
        f"{R[machine]}\t"
        f"{Q[machine]:.2f}"
    )

print("\nTotal Reward =", total_reward)
```
